```python
# 1

generator = pipeline("text-generation", model="distilgpt2")

outputs = generator(
    "Today we will learn about Transformers and",
    max_length=18,  
    num_return_sequences=2
)

for i, output in enumerate(outputs):
    print(f"Sentence {i+1}: {output['generated_text']}")
```

I used num_return_sequences=2 to generate two outputs and controlled the length using max_length.

```python
# 2
generator = pipeline("text-generation", model="distilgpt2")

result = generator(
    "Bonjour, aujourd'hui nous allons apprendre",
    max_length=25
)

print(result[0]["generated_text"])
```

I used the same text-generation pipeline but provided a non-English prompt to generate text in another language.

```python
# 3
unmasker_cased = pipeline("fill-mask", model="bert-base-cased")

results = unmasker_cased(
    "This course will teach you all about [MASK] models.",
    top_k=5
)

for r in results:
    print(r["sequence"])
```

For bert-base-cased, the mask token is MASK
The model predicts the most likely words to complete the sentence.

```python
# 4
pos_tagger = pipeline(
    "token-classification",
    model="vblagoje/bert-english-uncased-finetuned-pos",
    grouped_entities=True
)

sentence = "My name is Sylvain and I work at Hugging Face in Brooklyn."
pos_results = pos_tagger(sentence)

for item in pos_results:
    print(item["word"], "->", item["entity_group"])
```

This model assigns part-of-speech tags (like noun, verb, proper noun) to each word in the sentence.

```python
# 5
translator_fr = pipeline("translation_en_to_fr", model="Helsinki-NLP/opus-mt-en-fr")
translator_de = pipeline("translation_en_to_de", model="Helsinki-NLP/opus-mt-en-de")
translator_es = pipeline("translation_en_to_es", model="Helsinki-NLP/opus-mt-en-es")

sentence = "My name is Sylvain and I work at Hugging Face in Brooklyn."

print("French:", translator_fr(sentence)[0]["translation_text"])
print("German:", translator_de(sentence)[0]["translation_text"])
print("Spanish:", translator_es(sentence)[0]["translation_text"])
```

I used Helsinki-NLP translation models to translate the sentence into French, German, and Spanish.

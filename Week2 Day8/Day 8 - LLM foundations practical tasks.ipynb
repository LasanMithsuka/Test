````python
OPENAI_API_KEY = "OpenAIKey"
# Pass the key directly to the client initialization
client = OpenAI(base_url="https://ds-ai-internship.openai.azure.com/openai/v1/",
                api_key=OPENAI_API_KEY)
````

````python
def call_llm(prompt, system=None, model="gpt-4.1-mini", **kwargs):
    messages = []
    if system:
        messages.append({"role": "system", "content": system})
    messages.append({"role": "user", "content": prompt})

    resp = client.chat.completions.create(
        model=model,
        messages=messages,
        **kwargs
    )
    return resp.choices[0].message.content
````

#### 3. Call the OpenAI API with different `temperature` values

Prompt: "Write a two sentence story about a robot learning to cook."

`temperature` values: `0`, `0.7` and `1.5` three times for each setting.

Compare the outputs.

Tip: With Chat Completions API you can use parameter `n=3` to generate three different responses to your prompt with one call.

````python
prompt = "Write a two sentence story about a robot learning to cook."

for temp in [0, 0.7, 1.5]:
    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[{"role": "user", "content": prompt}],
        temperature=temp,
        n=3
    )

    print(f"Temperature = {temp}")
    for i, choice in enumerate(response.choices, 1):
        print(f"\nStory {i}:\n{choice.message.content}")
````

Low temperature (0) gives more predictable answers.
Higher temperature (1.5) makes the stories more creative and different.

#### 4. Call the OpenAI API with different `max_output_tokens`/`max_tokens` values.

Prompt: "Explain the concept of machine learning in simple terms."

`max_output_tokens` (Response API call)/`max_tokens` (Chat Completions API call) values: `50`, `100` and `200`.

Compare the length and detail of the responses.

````python
prompt = "Explain the concept of machine learning in simple terms."

outputs = {}

for tokens in [50, 100, 200]:
    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[{"role": "user", "content": prompt}],
        max_tokens=tokens
    )
    outputs[tokens] = response.choices[0].message.content
    print(f"\n max_tokens = {tokens} \n")
    print(outputs[tokens])
````

When max_tokens was 50, the explanation was short and very basic.
With 100 tokens, the answer became clearer and included a little more detail.
With 200 tokens, the explanation was more complete and sometimes included examples.

Increasing max_tokens allows the model to give longer and more detailed answers.

````python
prompt = "Write a list of 5 completely random and unusual hobbies that no one has ever heard of."

outputs = {}

for top_p in [0.3, 0.7, 1.0]:
    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[{"role": "user", "content": prompt}],
        temperature=0.9,
        top_p=top_p
    )
    outputs[top_p] = response.choices[0].message.content
    print(f"\n top_p = {top_p} \n")
    print(outputs[top_p])
````

### Task 3.1: Router Agent (Routing Pattern)

**Requirements:**
- Create a router that classifies content into: "social_media", "article", or "comment"
- Route should be based on length, structure, and style
- Return the classification decision

````python
# TODO: Implement Router Agent
# This agent analyzes content and classifies it

# Your code here:
def route_content_type(content: str) -> str:
    
    text = (content or "").strip()
    length = len(text)

    # Heuristics
    if length >= 450 or text.count("\n") >= 3:
        return "article"
    if length <= 140:
        return "comment"
    return "social_media"
````

### Task 3.2: Parallel Analysis Agents (Parallelization Pattern)

**Requirements:**
- Implement 3 agents that run in parallel:
  1. **Safety Checker**: Detect toxic, harmful, or inappropriate content
  2. **Tone Analyzer**: Assess sentiment (positive/negative/neutral) and professionalism
  3. **Grammar Checker**: Identify spelling, grammar, and clarity issues
- Each agent should return a structured assessment
- Execute them concurrently for efficiency

````python
# TODO: Implement Parallel Analysis Agents
# These agents analyze different aspects simultaneously

# Your code here:
import asyncio
from typing import List, Optional
from pydantic import BaseModel, Field

class SafetyAssessment(BaseModel):
    is_safe: bool = Field(..., description="True if content is safe to publish.")
    flags: List[str] = Field(default_factory=list, description="List of safety flags.")
    rationale: str = Field(..., description="Short explanation of the safety decision.")
    suggestion: Optional[str] = Field(None, description="Suggested safer rewrite approach if unsafe.")

class ToneAssessment(BaseModel):
    sentiment: str = Field(..., description="positive, neutral, or negative")
    professionalism: str = Field(..., description="professional, casual, or aggressive")
    tone_notes: List[str] = Field(default_factory=list, description="Short bullet notes about tone.")
    suggested_rephrase: Optional[str] = Field(None, description="Optional rephrase to make tone more constructive.")

class GrammarAssessment(BaseModel):
    issues: List[str] = Field(default_factory=list, description="List of grammar/spelling/clarity issues.")
    corrected_text: str = Field(..., description="A corrected version of the original content (grammar/spelling).")


async def safety_checker(content: str) -> SafetyAssessment:
    prompt = (
        "You are a content safety moderator.\n"
        "Assess the content for harmful/toxic/harassing content.\n"
        "Return a strict, structured result.\n\n"
        f"CONTENT:\n{content}"
    )
    model = llm.with_structured_output(SafetyAssessment)
    return await model.ainvoke(prompt)

async def tone_analyzer(content: str) -> ToneAssessment:
    prompt = (
        "You are a tone and sentiment analyzer.\n"
        "Classify sentiment (positive/neutral/negative) and professionalism (professional/casual/aggressive).\n"
        "Provide short tone notes and an optional constructive rephrase if needed.\n\n"
        f"CONTENT:\n{content}"
    )
    model = llm.with_structured_output(ToneAssessment)
    return await model.ainvoke(prompt)

async def grammar_checker(content: str) -> GrammarAssessment:
    prompt = (
        "You are a grammar and clarity checker.\n"
        "List grammar/spelling/clarity issues, then provide corrected text.\n"
        "Do not change meaning.\n\n"
        f"CONTENT:\n{content}"
    )
    model = llm.with_structured_output(GrammarAssessment)
    return await model.ainvoke(prompt)


async def run_parallel_analyses(content: str):
    safety_task = safety_checker(content)
    tone_task = tone_analyzer(content)
    grammar_task = grammar_checker(content)

    safety, tone, grammar = await asyncio.gather(safety_task, tone_task, grammar_task)
    return safety, tone, grammar
````

### Task 3.3: Evaluator Agent (Evaluator-Optimizer Pattern - Part 1)

**Requirements:**
- Aggregate results from the 3 parallel agents
- Calculate an overall content quality score (0-100)
- Make a decision: APPROVE (score ≥ 70) or REJECT (score < 70)
- For approved content, provide specific improvement suggestions

````python
# TODO: Implement Evaluator Agent
# This agent aggregates findings and makes decisions

# Your code here:
class EvaluationResult(BaseModel):
    score: int = Field(..., description="Overall quality score from 0 to 100.")
    decision: str = Field(..., description="APPROVE or REJECT.")
    reasons: List[str] = Field(default_factory=list, description="Reasons for the decision.")
    improvement_suggestions: List[str] = Field(default_factory=list, description="Specific improvements to apply if approved.")


def evaluate_content(
    safety: SafetyAssessment,
    tone: ToneAssessment,
    grammar: GrammarAssessment
) -> EvaluationResult:
    score = 100
    reasons = []
    suggestions = []

    if not safety.is_safe:
        score = max(0, score - 60)
        reasons.append("Safety: Content flagged as potentially harmful or toxic.")
        if safety.flags:
            reasons.append(f"Safety flags: {', '.join(safety.flags)}")
        if safety.suggestion:
            suggestions.append(safety.suggestion)

    if tone.professionalism.lower() == "aggressive":
        score -= 20
        reasons.append("Tone: Aggressive language detected.")
        if tone.suggested_rephrase:
            suggestions.append(f"Tone rephrase: {tone.suggested_rephrase}")
    elif tone.professionalism.lower() == "casual":
        score -= 5
        reasons.append("Tone: Casual tone (usually fine, may need polishing depending on context).")

    issue_count = len(grammar.issues)
    if issue_count > 0:
        penalty = min(25, issue_count * 5)
        score -= penalty
        reasons.append(f"Grammar: {issue_count} issue(s) found (penalty {penalty}).")
        suggestions.append("Fix grammar/spelling and improve clarity.")

    score = max(0, min(100, score))

    decision = "APPROVE" if score >= 70 and safety.is_safe else "REJECT"

    if decision == "APPROVE":
        if tone.tone_notes:
            suggestions.append("Tone polish: " + "; ".join(tone.tone_notes[:3]))

        suggestions.append("Use corrected version as baseline, then enhance readability.")

    return EvaluationResult(
        score=score,
        decision=decision,
        reasons=reasons,
        improvement_suggestions=list(dict.fromkeys(suggestions))
    )
````

### Task 3.4: Optimizer & Enhancer Agents (Prompt Chaining + Evaluator-Optimizer)

**Requirements:**
- **Optimizer Agent**: Generate specific improvements based on evaluator feedback
- **Enhancer Agent**: Apply improvements to create an enhanced version
- Implement as a chain: Original Content → Optimizer → Enhancer → Final Content
- (Optional) Add a re-evaluation loop if initial enhancement score is still low

````python
# TODO: Implement Optimizer and Enhancer Agents
# These agents improve content based on feedback

# Your code here:
class OptimizationPlan(BaseModel):
    changes: List[str] = Field(default_factory=list, description="Concrete improvements to apply.")
    target_style: str = Field(..., description="Target writing style based on content type.")

class EnhancedContent(BaseModel):
    enhanced_text: str = Field(..., description="Final improved content.")
    notes: List[str] = Field(default_factory=list, description="What was improved.")


async def optimizer_agent(content: str, content_type: str, evaluation: EvaluationResult) -> OptimizationPlan:
    prompt = (
        "You are a writing optimizer.\n"
        "Given the content type and evaluator feedback, propose a clear improvement plan.\n"
        "Make it specific (grammar fixes, tone adjustments, clarity, readability).\n\n"
        f"CONTENT TYPE: {content_type}\n"
        f"EVALUATOR SUGGESTIONS: {evaluation.improvement_suggestions}\n\n"
        f"ORIGINAL CONTENT:\n{content}"
    )
    model = llm.with_structured_output(OptimizationPlan)
    return await model.ainvoke(prompt)

async def enhancer_agent(content: str, content_type: str, grammar: GrammarAssessment, plan: OptimizationPlan) -> EnhancedContent:
    prompt = (
        "You are a content enhancer.\n"
        "Start from the corrected text, then apply the optimization plan.\n"
        "Keep original meaning. Make it suitable for the content type.\n"
        "Return only the enhanced content and short notes.\n\n"
        f"CONTENT TYPE: {content_type}\n"
        f"CORRECTED BASE TEXT:\n{grammar.corrected_text}\n\n"
        f"PLAN:\n{plan.changes}\n\n"
        "Produce the final enhanced version."
    )
    model = llm.with_structured_output(EnhancedContent)
    return await model.ainvoke(prompt)


async def enhancement_chain(
    original: str,
    content_type: str,
    safety: SafetyAssessment,
    tone: ToneAssessment,
    grammar: GrammarAssessment,
    evaluation: EvaluationResult,
    max_iters: int = 2
):
    
    if evaluation.decision != "APPROVE":
        return None, evaluation

    current_text = original
    current_eval = evaluation
    current_grammar = grammar
    notes = []

    for i in range(max_iters):
        plan = await optimizer_agent(current_text, content_type, current_eval)
        enhanced = await enhancer_agent(current_text, content_type, current_grammar, plan)

        notes.extend(enhanced.notes)
        current_text = enhanced.enhanced_text

        safety2, tone2, grammar2 = await run_parallel_analyses(current_text)
        current_eval = evaluate_content(safety2, tone2, grammar2)
        current_grammar = grammar2

        if current_eval.score >= 90:
            break

    return {
        "enhanced_text": current_text,
        "enhancement_notes": list(dict.fromkeys(notes)),
        "final_score": current_eval.score,
        "final_decision": current_eval.decision
    }, current_eval
````

### Task 3.5: Orchestrator (Orchestrator-Worker Pattern)

**Requirements:**
- Create a master orchestrator that coordinates the entire pipeline:
  1. Route content type
  2. Run parallel analysis
  3. Evaluate and decide
  4. If approved, optimize and enhance
  5. Return final result with metadata
- Handle both approval and rejection cases
- Provide clear logging of each step

````python
# TODO: Implement Orchestrator
# This coordinates the entire moderation pipeline

# Your code here:
from typing import Dict, Any

async def moderate_and_enhance(content: str) -> Dict[str, Any]:
    print("\n[Orchestrator] Starting moderation pipeline...")
    raw = (content or "").strip()

    content_type = route_content_type(raw)
    print(f"[Router] Classified as: {content_type}")

    print("[Parallel] Running safety, tone, and grammar checks...")
    safety, tone, grammar = await run_parallel_analyses(raw)

    print("[Evaluator] Aggregating results and scoring...")
    evaluation = evaluate_content(safety, tone, grammar)

    result: Dict[str, Any] = {
        "content_type": content_type,
        "analysis": {
            "safety": safety.model_dump(),
            "tone": tone.model_dump(),
            "grammar": grammar.model_dump(),
        },
        "evaluation": evaluation.model_dump(),
    }

    if evaluation.decision == "REJECT":
        print(f"[Decision] REJECT (score={evaluation.score})")

        feedback = {
            "message": "Your content was not approved. Please review the feedback and try again.",
            "reasons": evaluation.reasons,
            "suggestions": evaluation.improvement_suggestions,
            "constructive_rephrase": tone.suggested_rephrase or safety.suggestion
        }
        result["final"] = {"decision": "REJECT", "feedback": feedback}
        return result

    print(f"[Decision] APPROVE (score={evaluation.score})")
    print("[Enhancement] Running optimizer → enhancer chain...")
    enhanced_payload, final_eval = await enhancement_chain(
        original=raw,
        content_type=content_type,
        safety=safety,
        tone=tone,
        grammar=grammar,
        evaluation=evaluation,
        max_iters=2
    )

    result["final"] = {
        "decision": "APPROVE",
        "enhanced": enhanced_payload,
        "user_feedback": {
            "message": "Your content was approved and improved for clarity and tone.",
            "applied_suggestions": evaluation.improvement_suggestions
        }
    }
    return result
````

## 📊 Part 5: Reflection & Analysis

### Task 5.1: Pattern Usage Documentation

Document how you used each agentic pattern in your implementation.

### ✍️ YOUR PATTERN USAGE ANALYSIS

**1. Routing Pattern:**
- Where used: I used routing in the route_content_type() function at the beginning of the system. It checks the length and structure of the text and classifies it as social media, article, or comment.

- Why effective: This helps the system understand what kind of content it is dealing with. For example, a short comment should not be treated the same as a long article. It also helps later steps choose the right tone and improvements.

**2. Parallelization Pattern:**
- Where used: The function run_parallel_analyses() runs the safety checker, tone analyzer, and grammar checker at the same time using asyncio.gather.

- Why effective: These checks do not depend on each other, so they can run together instead of one after another.

- Performance benefit: Running them in parallel makes the system faster because all three checks finish roughly at the same time instead of waiting for each one separately.

**3. Evaluator-Optimizer Pattern:**
- Where used: The evaluator is implemented in evaluate_content(). It collects the results from the safety, tone, and grammar agents and calculates a score. After that, the optimizer suggests improvements.

- How feedback loop works: If the content is approved, the system improves the text and then checks it again. This helps make sure the final version is better than the original.

**4. Prompt Chaining Pattern:**
- Where used: Prompt chaining happens during the content improvement stage.

- Stages in chain: 
1. Original user content
2. Grammar correction
3. Optimization suggestions
4. Enhanced final version

**5. Orchestrator-Worker Pattern:**
- How orchestration is managed: The main function moderate_and_enhance() works as the orchestrator. It controls the whole process from classification to the final output.

- Worker coordination: Each worker (safety, tone, grammar, optimizer, enhancer) does one specific job. The orchestrator collects their results and decides what to do next.
